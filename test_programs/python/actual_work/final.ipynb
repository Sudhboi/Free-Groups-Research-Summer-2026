```python
def test_wh_automorphisms(partlist, w, fg):
    for p1, p2 in partlist:
        for A in [p1, p2]:
            for x in A:
                phi = generate_wh_automorphism(x, A, fg)
                nw = phi.map(w)
                if nw.length < w.length:
                    return phi, nw
    assert True, "This should not be happening"
```
