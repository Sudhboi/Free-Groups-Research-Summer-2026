```python
from graph_tool.all import *
from graph_tool.draw import *
import scipy
```

```python
arr = [[0, 8, 0, 2],
       [0, 0, 0, 0],
       [0, 0, 0, 5],
       [0, 0, 0, 0]]
g = Graph(scipy.sparse.lil_matrix(arr), directed=False)
```

```python
weight = g.edge_properties["weight"]
pos = sfdp_layout(g, eweight=weight)
graph_draw(g, edge_text=weight, pos=pos)
```

```python
mc, part = min_cut(g, weight)
mc
```

```python
graph_draw(g, edge_text=weight, vertex_fill_color=part)
```
